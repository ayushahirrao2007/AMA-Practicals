```python
# A* Search Algorithm

graph = {
    'A': {'B': 1, 'C': 3},
    'B': {'D': 3, 'E': 1},
    'C': {'F': 2},
    'D': {'G': 2},
    'E': {'G': 4},
    'F': {'G': 1},
    'G': {}
}

# Heuristic values
h = {
    'A': 5,
    'B': 4,
    'C': 4,
    'D': 2,
    'E': 3,
    'F': 1,
    'G': 0
}

start = 'A'
goal = 'G'

open_list = [start]
cost = {start: 0}
parent = {start: None}

while open_list:

    # Select node with minimum f(n)
    current = min(open_list, key=lambda x: cost[x] + h[x])

    if current == goal:
        break

    open_list.remove(current)

    for neighbour, edge_cost in graph[current].items():

        new_cost = cost[current] + edge_cost

        if neighbour not in cost or new_cost < cost[neighbour]:
            cost[neighbour] = new_cost
            parent[neighbour] = current
            open_list.append(neighbour)

# Create path
path = []
node = goal

while node is not None:
    path.append(node)
    node = parent[node]

path.reverse()

print("Path:", " -> ".join(path))
print("Cost:", cost[goal])
```
