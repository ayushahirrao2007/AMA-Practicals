```python
# Program to implement Bayes' Theorem

# P(A) = Probability of having a disease
P_A = 0.01

# P(B|A) = Probability of positive test when disease is present
P_B_given_A = 0.95

# P(B|not A) = Probability of positive test when disease is absent
P_B_given_not_A = 0.05

# P(not A)
P_not_A = 1 - P_A

# P(B) using Total Probability
P_B = (P_B_given_A * P_A) + (P_B_given_not_A * P_not_A)

# Bayes' Theorem
P_A_given_B = (P_B_given_A * P_A) / P_B

print("Probability of disease given positive test:",
      P_A_given_B)
```
